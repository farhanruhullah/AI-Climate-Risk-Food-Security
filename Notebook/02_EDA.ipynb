# Exploratory Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

df = pd.read_csv("../data/climate_food_security_FINAL.csv")
```

```python
trend = df.groupby("year")["rice_yield"].mean().reset_index()
fig = px.line(trend, x="year", y="rice_yield", markers=True, title="Rice Yield Trend")
fig.show()
```
